```python
%matplotlib inline
import matplotlib.pyplot as plt
import matplotlib

import os
import numpy as np

from sklearn.cluster import DBSCAN

from src.colors import colors_new_legend, generate_colors
```

```python
variables_path = "data/variables"
figures_path = "data/figures"
berenslab_data_path = "data/embeddings"
```

```python
tsne = np.load(os.path.join(variables_path, "tsne_reparsed.npy"))
```

### DBSCAN: Test different params 

```python
epsilons = [0.5,1,1.5,2,2.5,3,3.5,4,4.5,5]
num_samples = [2,3,5,10,15,20,30,50,100]

for eps in epsilons:
    for num_sample in num_samples:
        print(f'=====> eps={eps},min_samples={num_sample}')
        
        dbscan = DBSCAN(eps=eps, min_samples=num_sample).fit(tsne)
        
        labels = dbscan.labels_
        print('num of clusters',len(np.unique(dbscan.labels_)))
        if len(np.unique(dbscan.labels_))<80 or len(np.unique(dbscan.labels_))>500:
            continue
        
        colors_new_legend = generate_colors(len(np.unique(dbscan.labels_)))
        colors = [colors_new_legend[i] for i in labels]
        
        fig, ax = plt.subplots(figsize=(5.5, 5.5), dpi=200, layout="constrained")
        ax.scatter(
            tsne[:, 0],
            tsne[:, 1],
            s=0.5,
            c=colors,
            marker=".",
            #     linewidths=None,
            #     ec="None",
            #     alpha=0.01,
            #     rasterized=True,
        )
        ax.axis("equal")
        ax.grid()
        # ax.xticks([])
        # ax.yticks([])
        ax.axis("off")
        plt.show()
        # fig.savefig(os.path.join(figures_path, "tsne_bert_reparsed.png"))
        
```

## DBSCAN with the best params eps=2.5, min_samples=15

```python
dbscan = DBSCAN(eps=2.5, min_samples=15).fit(tsne)
```

```python
labels = dbscan.labels_
colors_new_legend = generate_colors(len(np.unique(dbscan.labels_)))
colors = [colors_new_legend[i] for i in labels]
```

```python
print(f"len of labels:{len(set(labels))}")
```

```python
fig, ax = plt.subplots(figsize=(5.5, 5.5), dpi=200, layout="constrained")

ax.scatter(
    tsne[:, 0],
    tsne[:, 1],
    s=0.5,
    c=colors,
    marker=".",
)

ax.set_title("tsne+dbscan")
ax.axis("equal")
ax.grid()
ax.axis("off")
plt.show()

fig.savefig('data/results/tsne-dbscan.png')
```
