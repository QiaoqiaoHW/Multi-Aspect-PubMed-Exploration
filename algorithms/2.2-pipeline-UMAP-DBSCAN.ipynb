```python
%matplotlib inline
import matplotlib.pyplot as plt
import matplotlib

import os
import numpy as np

from sklearn.cluster import DBSCAN

from src.colors import colors_new_legend, generate_colors
```

```python
variables_path = "data/variables"
figures_path = "data/figures"
berenslab_data_path = "data/embeddings"
```

```python
umap_embeddings = np.load(os.path.join(variables_path, "umap_reparsed.npy"))
```

### DBSCAN: Test different params 

```python
epsilons = [0.02*i for i in range(5,10)]
num_samples = [5*i for i in range(1,6) ]

for eps in epsilons:
    for num_sample in num_samples:
        print(f'=====> eps={eps},min_samples={num_sample}')
        
        dbscan = DBSCAN(eps=eps, min_samples=num_sample).fit(umap_embeddings)
        
        labels = dbscan.labels_
        print('num of clusters',len(np.unique(dbscan.labels_)))
        if len(np.unique(dbscan.labels_))<15 or len(np.unique(dbscan.labels_))>300:
            continue
        
        colors_new_legend = generate_colors(len(np.unique(dbscan.labels_)))
        colors = [colors_new_legend[i] for i in labels]
        
        fig, ax = plt.subplots(figsize=(5.5, 5.5), dpi=200, layout="constrained")
        ax.scatter(
            umap_embeddings[:, 0],
            umap_embeddings[:, 1],
            s=0.5,
            c=colors,
            marker=".",
        )
        ax.axis("equal")
        ax.grid()
        ax.axis("off")
        plt.show()
        
```

## DBSCAN with the best params eps=0.12, min_samples=25

```python
# dbscan = DBSCAN(eps=0.1, min_samples=15).fit(umap_embeddings)
dbscan = DBSCAN(eps=0.12, min_samples=25).fit(umap_embeddings)
```

```python
labels = dbscan.labels_
print(np.unique(dbscan.labels_))

colors_new_legend = generate_colors(len(np.unique(dbscan.labels_)))
colors = [colors_new_legend[i] for i in labels]
```

```python
fig, ax = plt.subplots(figsize=(5.5, 5.5), dpi=200, layout="constrained")

ax.scatter(
    umap_embeddings[:, 0],
    umap_embeddings[:, 1],
    s=0.5,
    c=colors,
    marker="."
)
ax.set_title("umap+dbscan")
ax.axis("equal")
ax.grid()
ax.axis("off")
plt.show()

fig.savefig('data/results/umap-dbscan.png')
```
